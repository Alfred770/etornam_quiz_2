```python
from sklearn.linear_model import LinearRegression
hours_studied = [[1], [2], [3], [4], [5]]
exam_scores = [50, 60, 70, 80, 90]
model = LinearRegression()
model.fit(hours_studied, exam_scores)

predicted_score = model.predict([[8]])
print(f"Predicted exam score for 8 hours of study: {predicted_score[0]}")
print(f"slope: {model.coef_[0]}, intercept: {model.intercept_}")
```

```python
hours_studied=[1,2,3,4,5]
exam_scores=[40,50,60,70,80]
print("Hours Studied:", hours_studied)
print("Exam Scores:",exam_scores)
```

```python
print("Training data:", hours_studied)
print("Exam scores:", exam_scores)
print("Slope:", model.coef_[0])
print("Intercept:", model.intercept_)
print("Prediction for 8hours:",model.predict([[8]])[0])
```
